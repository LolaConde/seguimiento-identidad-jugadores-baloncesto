# **Justificación del uso de un modelo entrenado para baloncesto**

Para analizar el impacto del modelo de detección se compara OC-SORT utilizando el detector entrenado específicamente para jugadores de baloncesto con OC-SORT utilizando el modelo YOLO versión 26 de Ultralytics preentrenado en el dataset COCO (se utiliza la clase persona). En ambos casos, se mantienen fijos los hiperparámetros seleccionados (*seleccion_hiperparametros_OCSORT.ipynb*). La comparación se evalúa en el conjunto de test, y los resultados se muestran en la siguiente tabla.

<p><em>Valores de métricas para OC-SORT con un modelo de detección genérico y un modelo entrenado para baloncesto</em></p>

El detector genérico obtiene mayor número de IDSW (812 frente a 476), y crea un mayor número de IDs que el modelo entrenado para baloncesto (945 frente a 421), lo que podría deberse a que detecta árbitros, público, y otras personas que no son jugadores, y a que no detecta a los jugadores de forma tan consistente, lo que fragmenta las trayectorias y genera un mayor número de IDs e intercambios de identidad.

Por otra parte, el número de IDTP es mucho mayor en el modelo reentrenado, y menor en número de IDFN, debido a su mejor detección de jugadores. Sin embargo, hay mayor número de IDFP en el modelo reentrenado (41237 frente a 25950), lo que podría deberse a que, al detectar de forma más consistente a los jugadores, el tracker mantiene las trayectorias durante más frames, por lo que tras un intercambio de identidad, las trayectorias pueden prolongarse durante un mayor número de frames. Aunque el número de IDFP es mayor en el modelo entrenado para baloncesto, el resto de métricas son mucho mejores, por lo que el modelo tiene mejor consistencia en el seguimiento.

Se muestra que un modelo entrenado específicamente para detectar jugadores de baloncesto, que vea los jugadores desde los distintos ángulos que suelen tener los vídeos de los partidos, funciona mejor que un modelo genérico para detectar personas, resaltando la importancia de un buen modelo de detección de jugadores.
